converting time stamps according to the sun strength

```python
df['Time_Hour'] = pd.to_datetime(df['Time']).dt.hour
```

```python
conditions = [
    (df['Time_Hour'] >= 7) & (df['Time_Hour'] < 11) | (df['Time_Hour'] >= 16) & (df['Time_Hour'] < 19),
    (df['Time_Hour'] >= 11) & (df['Time_Hour'] < 16)
]
choices = [2, 1]
df['Time_Category'] = np.select(conditions, choices, default=0)
```
